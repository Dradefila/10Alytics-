```python

relative_paths_file = current_working_dir / 'relative_paths.txt' # relative path txt
absolute_paths_file = current_working_dir / 'absolute_paths.txt' # absolute path txt

# Collect relative and absolute paths of subdirectories and files in 'alphabet'
relative_paths = []
absolute_paths = []

for item in alphabet_dir.rglob('*'):  # Recursively get all files and subdirectories
    relative_paths.append(item.relative_to(current_working_dir))
    absolute_paths.append(item.resolve())

# Write to relative_paths.txt
with relative_paths_file.open('w') as rel_file:
    for path in relative_paths:
        rel_file.write(f"{path}\n")

# Write to absolute_paths.txt
with absolute_paths_file.open('w') as abs_file:
    for path in absolute_paths:
        abs_file.write(f"{path}\n")
```

```python
def list_existed_files(persons_dir, filenames):
    """
    List existing file names in the given directory (case-sensitive) with their absolute paths.

    Parameters:
    - persons_dir (Path): Path object pointing to the directory to search in.
    - filenames (list): List of filenames to check for existence.

    Returns:
    - tuple: (existing_filenames, absolute_paths)
    """
    # Ensure persons_dir is a Path object
    persons_dir = Path(persons_dir)
    
    # Get a set of files currently in the directory (case-sensitive)
    existing_files_in_dir = set(persons_dir.iterdir())

    # Initialize results
    existing_filenames = []
    absolute_paths = []

    # Check for each filename
    for filename in filenames:
        filepath = persons_dir / filename
        if filepath in existing_files_in_dir:  # Check existence and case sensitivity
            existing_filenames.append(filename)
            absolute_paths.append(filepath.resolve())

    return existing_filenames, absolute_paths

# Example usage
# Define current working directory and subdirectory
current_working_dir = Path.cwd()  # Replace with the actual working directory if needed
persons_dir = current_working_dir / "Persons"

# Ensure the subdirectory exists
persons_dir.mkdir(exist_ok=True)

print(f"Subdirectory created at: {persons_dir}")

# List of filenames to check
filenames = [
    'Alfred Nobel.txt', 'Max Planck.txt', 'Alan Turing.csv',
    'Henry-Cavil.txt', 'Cameron Diaz.txt', 'Isaac Newton.txt', 
    'Leonardo DiCaprio.json', 'Natalie portman.txt' 
]

# Call the function
existing_filenames, absolute_paths = list_existed_files(persons_dir, filenames)

# Output the results
print("Existing Filenames:", existing_filenames)
print("Absolute Paths:", absolute_paths)
```

```python

from shutil import move #This is a standard library module in Python that provides high-level file operations. It offers functions for copying, moving, deleting, and compressing files and directories
```

```python
def move_corrupted_files(persons_dir, persons_corrupted_dir):
    """
    Move all corrupted files from persons_dir to persons_corrupted_dir.
    """
    persons_dir = Path(persons_dir)  # Ensure persons_dir is a Path object
    persons_corrupted_dir = Path(persons_corrupted_dir)  # Ensure persons_corrupted_dir is a Path object
    
    # This is to get all the list of corrupted files. i.e. the files not ending with the right format .txt
    corrupt_files = list_corrupted_files(persons_dir)
    
    # This is for moving each corrupted file to the new directory i.e. persons_corrupted_dir in Persons_corrupted 
    for corrupted_file in corrupt_files:
        source = persons_dir / corrupt_file
        destination = persons_corrupted_dir / corrupt_file
        move(str(source), str(destination))  # Move the file
        print(f"Moved {source} to {destination}") # nothing to move because no corrupted file
```
