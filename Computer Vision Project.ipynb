```python
# I am importing OpenCV 

import cv2
```

```python
# I am defining a variable called face_cascade
# to extract the frontal face default classifier 

face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml') 
```

```python
# Create a fuction that will detect the face

def detect(gray, frame): # the gray scale of the picture/image, frame is the color scale
    face = face_cascade.detectMultiScale(gray, 1.3, 5) # the measurement
    for (x, y, w, h) in face: # x-axis, y-axis, width and height == x, y, w, h,
        cv2.rectangle(frame, (x,y), (x+w, y+h), (0,100,200), 4) # 255 is the color code, 3 is the thickness of the line for the detection
        roi_gray = gray[y:y+h, x:x+w]
        roi_color = frame[y:y+h, x:x+w] 
    return frame
```

```python
# using python code to turn up my camera
# you need excellent illumination
capture = cv2.VideoCapture(0) # put a 0 if you are using a local webcam or 1 if it is an exteranl webcam

while True:
    _, frame = capture.read() # to switch on the camera
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY) #converting the picture to gray scale
    canvas = detect(gray, frame) # to return frame
    cv2.imshow("Video", canvas) # to show the frame
    if cv2.waitKey(1) and 0xff == ord('q'): # to turn off my camera
        break 
capture.release()
cv2.destroyAllWindows() # to close all windows created
```
