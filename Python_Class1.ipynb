```python
##- What is the total cost incured from buying a bag of $100, phone of $500 and a power_bank of $150.

phone = int(input('what is the phone price? '))
bag = int(input('what is the bag price? '))
power_bank = int(input('what is the power bank price? '))

total_cost = phone + bag + power_bank
total_cost

```

```python
## Your Monthly income is $3,500 and you pay $750 in taxes. Calculate your disposable income by subtracting the tax from your income.

monthly_income = int(input('what is my monthly income? '))
tax = int(input('what is my tax? '))
disposable_income = monthly_income - tax
disposable_income
```

```python
## - You plan to forecast using the Simple Liner Forecast (Y = a + bx). What is Y? when a=1200, b=0.56 and x = 2500

a = int(input('what is the value for a? '))
b = float(input('what is the value for b? '))
x = int(input('what is the value for x? '))
Y = a + (b*x)
Y 
```
