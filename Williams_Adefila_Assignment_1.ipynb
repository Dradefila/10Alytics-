````python
def list_existed_files(persons_dir, filenames):
    # TODO: Write your code below

    new_dir = Path(persons_dir)
    persons_list=os.listdir(new_dir)

    existing_filenames = []
    absolute_paths = []

    for filename in filenames:
        if filename in persons_list:
            existing_filenames.append(filename)
            absolute_paths.append(new_dir / filename)
    return existing_filenames, absolute_paths
````
