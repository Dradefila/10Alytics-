```python
import seaborn as sns 
import matplotlib.pyplot as plt
sns.histplot(df1.age, edgecolor = 'k', linewidth = 2, color = 'thistle')
plt.show()
```

```python
# Create a figure and axes for the subplots
fig, axes = plt.subplots(3, 3, figsize=(15, 15))  # 3 rows, 3 columns

# Example variables (replace with your actual column names)
variables = ['age', 'lcavol', 'lweight', 'lbph', 'lcp', 
             'svi', 'gleason', 'pgg45', 'lpsa']

# Loop through the variables and create a histogram for each
for i, var in enumerate(variables):
    row = i // 3  # Get the row index
    col = i % 3  # Get the column index
    sns.histplot(df1[var], ax=axes[row, col], edgecolor='k', linewidth=1, color='blue')
    axes[row, col].set_title(var)  # Set the title for each subplot

plt.tight_layout()  # Adjust layout to prevent overlapping
plt.show()
```

```python
# Create a figure and axes for the subplots
fig, axes = plt.subplots(3, 3, figsize=(15, 15))

# Example variables (replace with your actual column names)
variables = ['age', 'lcavol', 'lweight', 'lbph', 'lcp',
             'svi', 'gleason', 'pgg45', 'lpsa']

# Loop through the variables and create scatter plots
for i, var1 in enumerate(variables):
    row = i // 3
    col = i % 3

    # Create scatter plots against 'lpsa' (or any other chosen variable)
    sns.scatterplot(x=var1, y='lpsa', data=df1, ax=axes[row, col], edgecolor='k', linewidth=0.5, color='orange')

    axes[row, col].set_title(f'{var1} vs. lpsa')  # More informative title
    axes[row, col].set_xlabel(var1)  # Label the x-axis
    axes[row, col].set_ylabel('lpsa')  # Label the y-axis

plt.tight_layout()
plt.show()
```

```python
# Calculate the correlation matrix
correlation_matrix = df1[variables].corr()  # Use the 'variables' list

# Print the correlation matrix or specific correlations
print("Correlation Matrix:")
print(correlation_matrix)

# To check correlation with 'lpsa' specifically:
correlation_with_lpsa = correlation_matrix['lpsa'].sort_values(ascending=False)
print("\nCorrelation with lpsa:")
print(correlation_with_lpsa)
```

```python
# Calculate the correlation matrix
correlation_matrix = df1[variables].corr()  # Use the 'variables' list

# Print the correlation matrix or specific correlations
print("Correlation Matrix:")
print(correlation_matrix)

# To check correlation with 'lpsa' specifically:
correlation_with_lpsa = correlation_matrix['lpsa'].sort_values(ascending=False)
print("\nCorrelation with lpsa:")
print(correlation_with_lpsa)

# --- Plotting the Correlation Heatmap ---
plt.figure(figsize=(10, 8))  # Adjust figure size as needed
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```
